# Tutorial: Conda Environment Installation and Setup

## What is a Virtual Environment?
A virtual environment is an isolated space on your computer. It allows you to install specific versions of Python and packages for a project without interfering with other projects or your operating system.


## Step 0: Download and Install Conda

Before creating an environment, you need Conda installed on your system. Choose the option that best fits your needs:

| Installer| Description |
| :--- | :--- |
| **Miniconda (Recommended)** | A lightweight version that installs only Conda, Python, and a few essential packages. You get a clean start for every project, installing only the packages you actually need. |
| **Anaconda** | Installs conda, Python, Anaconda Navigator, and hundreds of popular data science and machine learning packages into a pre-built base environment.|


### Installation Steps:
1. **Download the installer:** Go to the official [Anaconda](https://www.anaconda.com/download) site and download the installer for your OS (Windows, macOS, or Linux). More details can be found in the official [installation documentation](https://www.anaconda.com/docs/getting-started/installation).

For Linux users, you can download and install Miniconda directly via the terminal:

```bash
    curl -O https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh
```

Install Miniconda by running the following command:

```bash
    bash ~/Miniconda3-latest-Linux-x86_64.sh
```

2. **Agree the terms:** Enter `yes` to agree to the EULA.
3. **Choose location:** Press Enter to accept the default install location `(PREFIX=/home/<USER>/miniconda3)`, or enter another file path to specify an alternate installation directory. The installation might take a few minutes to complete.
4. **Initialize Conda:** Choose `yes` when asked if you want to initialize conda. This modifies your shell configuration to initialize conda whenever you open a new shell and to recognize conda commands automatically.
6. **Refresh your terminal:** Close and re-open your terminal window for the installation to fully take effect, or use the following command to refresh the terminal:

```bash
    source ~/.bashrc
```
### Verify your installation

Verify your installation by opening your terminal and running: 

```bash
    conda list
```
## Step 1: Create a New Environment

### Python

* To create a new environment, use the `conda create` command. It is best practice to give it a descriptive name and specify the Python version you need.

```bash
    conda create --name my_project_name python=X.X
```
> **Note:** Replace `X.X` with your desired Python version (e.g., `3.10`). When prompted `Proceed ([y]/n)?` type `y` and press **Enter** to confirm.  
> **Example:** `conda create --name data_analysis python=3.10` 

### R

* To create a new environment, use the `conda create` command. It is best practice to give it a descriptive name and specify the R version you need.

```bash
    conda create --name my_project_name r-base=X.X
```
> **Note:** Replace `X.X` with your desired R version (e.g., `4.3`) or simply type r-base to install the latest version. When prompted `Proceed ([y]/n)?` type `y` and press **Enter** to confirm.  
> **Example:** `conda create --name data_analysis r-base=4.3` 


## Step 2: Activate the Environment

Once created, you need to activate the environment to start working inside it:

```bash
    conda activate my_project_name
```

> **Note:** You can verify it, your terminal prompt will change from (base) to `my_project_name`. Anything you install or run now will stay within this isolated environment.  
> **Example:** `conda activate data_analysis` 


## Step 3: Install Packages

### Python

* **Install a single package:** With your environment active, you can install the specific libraries required for your project (e.g., Pandas, NumPy). To install a single package in python, run the following command:

```bash
    conda install <PACKAGE>
```

* **Install multiple packages:** list them separated by a space:

```bash
    conda install <PACKAGE> <PACKAGE> <PACKAGE>
```

> **Note:** Replace `PACKAGE` with the name of the package you want to install. If a package is not available in the Conda channels, you can run `pip install package_name` inside the active environment.  
> **Examples:**  
>`conda install pandas`  
>`conda install pandas numpy matplotlib`


* **Install a specific version:** By default conda installs the latest possible versions of the requested package and their dependencies that are compatible with the environment.

```bash
    conda install PACKAGE = X.X.X
```
> **Note:** Replace `PACKAGE` with the name of the package you want and `X.X.X` the version you need.  
> **Example:** `conda install pandas = 2.0.3`


* **Install packages in a different environment:** To install a package in another environment that is not your current active one, specify the environment name:
    
```bash
    conda install <PACKAGE> --name <ENVIRONMENT>
```

> **Note:** Replace `PACKAGE` with the name of the package you want to install and replace `ENVIRONMENT` with the name of the environment you want to install the package.  
> **Example:** `conda install numpy --name spatial_analysis` 


### R

When you are inside your active environment, instead of using `conda install <package>` (which defaults to searching for Python packages), you should use the `conda-forge` channel and the `r-` prefix: 

* **Install a single package (via conda):** To install R packages via Conda run the following command:

```bash
    conda install -c conda-forge r-<PACKAGE>
```


* **Install multiple packages (via conda):** When you are inside your active environment, instead of using `conda install <package>` (which defaults to searching for Python packages), you should use the conda-forge channel and the r- prefix: 

```bash
    conda install -c conda-forge r-<PACKAGE> r-<PACKAGE> r-<PACKAGE>
```


* **Install a specific version (via conda):**  By default conda installs the latest possible versions of the requested package and their dependencies that are compatible with the environment. To install R packages in a specific version via Conda, prefix the package name with `r-`:

```bash
    conda install r-PACKAGE = X.X
```
> **Note:** Replace `PACKAGE` with the name of the package you want and X.X the version you need.  
> **Example:** `conda install r-ggplot2=3.4.2`



* **Install R packages in a different environment (via Conda):** To install an R package in another environment that is not your current active one, specify the environment name:
    
```bash
    conda install r-PACKAGE --name <ENVIRONMENT>
```

> **Note:** Replace `PACKAGE` with the name of the R package you want to install (remembering the r- prefix) and ENVIRONMENT with the name of the target environment.  
> **Example:** `conda install r-ggplot2 --name spatial_analysis` 


**Recommended alternative inside R:** Once your conda environment is active and you are inside an R session, you can install packages the usual way.

* **Install a single package (in R):** To install R packages in R run the following command:

```r
    install.packages("PACKAGE")
```

* **Install multiple packages (in R):** list them separated by a space.

```r
    install.packages(c("PACKAGE", "PACKAGE","PACKAGE"))
```
> **Note:** Replace `PACKAGE` with the name of the package you want to install.   
> **Examples:**  
> `install.packages("ggplot2")`  
> `install.packages(c("ggplot2", "dplyr", "tidyr"))`

* **Install a specific version (in R):** To install a package in a specific version directly inside R, you can use the `install_version()` function from the `remotes` package (since base R does not support version control during installation):

```r
    remotes::install_version("PACKAGE", version = "X.X.X")
```

> **Note:** Replace PACKAGE with the name of the package and X.X.X with the exact version number you need. You will need to have the remotes package installed first `(install.packages("remotes"))`.  
> `remotes::install_version("ggplot2", version = "3.4.0")`


* **Install packages in a different environment (inside R):** To install a package into a specific library path or environment folder while inside an R session, use the `lib` argument:

```r
    install.packages("PACKAGE", lib = "PATH_TO_ENVIRONMENT_LIBRARY")
```
> **Note:** Replace `PACKAGE` with the name of the package and `PATH_TO_ENVIRONMENT_LIBRARY` with the file path to the target environment's library folder. 
> **Example:** `install.packages("ggplot2", lib = "/home/user/miniconda3/envs/spatial_analysis/lib/R/library")` 


## Step 4: Export and Share Environments (Optional)

If you want to share your project with others or back up your setup, you can export all installed packages and their exact versions into a configuration file (`environment.yml`).

* **Export your environment:**
  Make sure your environment is active, then run:
  
```bash
    conda env export > environment.yml
```

### Useful Management Commands

* Deactivate the environment when you want to return to your base system:

```bash
    conda deactivate 
```

* List all created environments:

```bash
    conda env list
```

* To remove an environment:

```bash
    conda env remove --name my_project_name
```

* To keep conda updated:

```bash
    conda update conda
```
