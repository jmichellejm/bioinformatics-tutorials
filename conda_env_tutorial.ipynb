# Tutorial: Conda Environment Installation and Setup

## 💡 What is a Virtual Environment?
A virtual environment is an isolated space on your computer. It allows you to install specific versions of Python and packages for a project without interfering with other projects or your operating system.

## 📥 Step 0: Download and Install Conda

Before creating an environment, you need Conda installed on your system. Choose the option that best fits your needs:

| Installer | Description |
| :--- | :--- |
| **Miniconda (Recommended)** | A lightweight version that installs only Conda, Python, and a few essential packages. You get a clean start for every project, installing only the packages you actually need. |
| **Anaconda** | Installs conda, Python, Anaconda Navigator, and hundreds of popular data science and machine learning packages into a pre-built base environment. |

### 🛠️ Installation Steps:
1. **Download the installer:** Go to the official [Anaconda](https://www.anaconda.com/download) site and download the installer for your OS (Windows, macOS, or Linux) . More details can be found in the official [installation documentation](https://www.anaconda.com/docs/getting-started/installation).

For Linux users, you can download and install Miniconda directly via the terminal :

```bash
    curl -O https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh
```

Install Miniconda by running the following command :

```bash
    bash ~/Miniconda3-latest-Linux-x86_64.sh
```

2. **Agree to terms:** Enter `yes` to agree to the EULA.
3. **Choose location:** Press Enter to accept the default install location `(PREFIX=/home/<USER>/miniconda3)`, or enter another file path to specify an alternate installation directory. The installation might take a few minutes to complete.
4. **Initialize Conda:** Choose `yes` when asked if you want to initialize conda. This modifies your shell configuration to initialize conda whenever you open a new shell and to recognize conda commands automatically.
5. **Refresh your terminal:** Close and re-open your terminal window for the installation to fully take effect, or use the following command to refresh the terminal:

```bash
    source ~/.bashrc
```

### 🔍 Verify your installation

Verify your installation by opening your terminal and running: 

```bash
    conda list
```


## 🛠️ Step 1: Create a New Environment

### 🐍 Python

* To create a new environment, use the `conda create` command. It is best practice to give it a descriptive name and specify the Python version you need.

```bash
    conda create --name my_project_name python=X.X
```
> **Note:** Replace `X.X` with your desired Python version (e.g., `3.10`). When prompted `Proceed ([y]/n)?` type `y` and press **Enter** to confirm.  
>
> **Example:** `conda create --name data_analysis python=3.10` 

### 📊 R

* To create a new environment, use the `conda create` command. It is best practice to give it a descriptive name and specify the R version you need.

```bash
    conda create --name my_project_name r-base=X.X
```
> **Note:** Replace `X.X` with your desired R version (e.g., `4.3`) or simply type `r-base` to install the latest version. When prompted `Proceed ([y]/n)?` type `y` and press **Enter** to confirm.  
>
> **Example:** `conda create --name data_analysis r-base=4.3` 


## ⚡ Step 2: Activate the Environment

Once created, you need to activate the environment to start working inside it:

```bash
    conda activate my_project_name
```

> **Note:** You can verify it because your terminal prompt will change from `(base)` to `my_project_name`. Anything you install or run now will stay within this isolated environment.  
>
> **Example:** `conda activate data_analysis` 


## 📦 Step 3: Install Packages

### 🐍 Python

* **Install a single package:** With your environment active, you can install specific libraries required for your project.

```bash
    conda install <PACKAGE>
```
>
> **Note:** Replace `PACKAGE` with the name of the package.  
>
> **Example:** `conda install pandas`

* **Install multiple packages:** List them separated by a space.

```bash
    conda install <PACKAGE> <PACKAGE> <PACKAGE>
```
>
> **Note:** Separate multiple package names with spaces.  
>
> **Example:** `conda install pandas numpy matplotlib`

* **Install a specific version:** By default, conda installs the latest compatible versions.

```bash
    conda install PACKAGE=X.X.X
```
>
> **Note:** Replace `PACKAGE` with the package name and `X.X.X` with the required version.  
>
> **Example:** `conda install pandas=2.0.3`

* **Install in a different environment:** Specify the environment name via `--name`.
    
```bash
    conda install <PACKAGE> --name <ENVIRONMENT>
```
>
> **Note:** Replace `PACKAGE` and `ENVIRONMENT` with your specific target names.  
>
> **Example:** `conda install numpy --name spatial_analysis` 


### 📊 R (via Conda)

* **Install a single package (via conda):** Use the `conda-forge` channel and the `r-` prefix.

```bash
    conda install -c conda-forge r-<PACKAGE>
```
>
> **Note:** Remember to include the `r-` prefix for R packages.  
>
> **Example:** `conda install -c conda-forge r-ggplot2`

* **Install multiple packages (via conda):** List them separated by spaces.

```bash
    conda install -c conda-forge r-<PACKAGE> r-<PACKAGE>
```
>
> **Note:** Ensure each R package has the `r-` prefix.  
>
> **Example:** `conda install -c conda-forge r-ggplot2 r-dplyr`

* **Install a specific version (via conda):**  By default conda installs the latest possible versions of the requested package and their dependencies that are compatible with the environment. To install R packages in a specific version via Conda, prefix the package name with `r-` prefix:


```bash
    conda install r-PACKAGE=X.X.X
```
>
> **Note:** Replace `PACKAGE` with the name and `X.X.X` with the version.  
>
> **Example:** `conda install r-ggplot2=3.4.2`

* **Install in a different environment (via conda):** Target a specific environment.
    
```bash
    conda install r-PACKAGE --name <ENVIRONMENT>
```
>
> **Note:** Specify both the `r-` package and target environment name.  
>
> **Example:** `conda install r-ggplot2 --name spatial_analysis` 


### 📊 R (Inside an R Session)

**Recommended alternative:** Once your environment is active and you are inside an R session, you can use standard R commands.

* **Install a single package:** Run `install.packages()`.

```r
    install.packages("PACKAGE")
```
>
> **Note:** Enclose the package name in quotes.  
>
> **Example:** `install.packages("ggplot2")`

* **Install multiple packages:** Use a character vector with `c()`.

```r
    install.packages(c("PACKAGE", "PACKAGE"))
```
>
> **Note:** Pass package names as a vector.  
>
> **Example:** `install.packages(c("ggplot2", "dplyr", "tidyr"))`

* **Install a specific version:** Use the `remotes` package.

```r
    remotes::install_version("PACKAGE", version = "X.X.X")
```
>
> **Note:** Requires `remotes` to be installed first (`install.packages("remotes")`).  
>
> **Example:** `remotes::install_version("ggplot2", version = "3.4.0")`

* **Install in a different environment library:** Use the `lib` argument.

```r
    install.packages("PACKAGE", lib = "PATH_TO_ENVIRONMENT_LIBRARY")
```
>
> **Note:** Point `lib` to the target library folder.  
>
> **Example:** `install.packages("ggplot2", lib = "/home/user/miniconda3/envs/spatial_analysis/lib/R/library")` 


## 📥 Step 4: Recreate and Clone an Environment from a Project

If you want to clone or download an existing project and replicate its exact environment with the same packages and versions, you can do it depending on the configuration files provided:

### 🐍 Conda (`environment.yml`)
If the project includes an `environment.yml` file, you can automatically recreate the environment with the exact Python/R version and all its dependencies:

```bash
    conda env create -f environment.yml
```

### 🐍 Python Pip (`requirements.txt`)

If the project shares its libraries through a standard Python requirements.txt file:

```bash
    conda create --name my_project_name python=X.X
    conda activate my_project_name 
    pip install -r requirements.txt
```

>
> **Note:** Always make sure to create and activate a clean environment before installing the requirements.

### 📊 R (renv package)

In professional R projects, the `renv` package is standard for locking specific package versions in an `renv.lock` file. Follow these steps to restore the environment:

1. **Open the project:** FInd the `.Rproj` file in your project folder and open it in RStudio.
2. **Open the R Console:** Navigate to the Console panel in RStudio.
3. **Restore dependencies:** Run the following command:

```r
    renv::restore()
```

>
> **Note:** This automatically downloads and installs the exact package versions registered in the renv.lock file into an isolated project library without affecting your global system.

💡 **Tip**: If the repository does not provide any configuration files (`renv.lock`, `environment.yml, requirements.txt`), check the project's `README.md` file for instructions, or inspect the script files to manually install the required packages using `install.packages()`.

## 📤 Step 5: Export and Share Environments (Optional)

If you want to share your project with others or back up your setup, you can export all installed packages and their exact versions into a configuration file (`environment.yml`).

* **Export your environment:**
  Make sure your environment is active, then run:
  
```bash
    conda env export > environment.yml
```

## ⚙️ Useful Management Commands

* **Deactivate** the environment when you want to return to your base system:

```bash
    conda deactivate 
```

* **List** all created environments:

```bash
    conda env list
```

* **Remove** an environment:

```bash
    conda env remove --name my_project_name
```

* **Update** conda:

```bash
    conda update conda
```
